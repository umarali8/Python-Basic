## Dictionary comprehension 

```python
# Syntax --> {Key:value for vars in iterable}
#print 1st 10 number and their square
{i:i**2 for i in range(1,11)}
```

```python
distances={'Karachi':1000,'dadu':2000,'sukkur':3000}
print(distances.items())
```

```python
#Using existing dict
distances={'Karachi':1000,'dadu':2000,'sukkur':3000}
{key:value*0.63 for(key,value) in distances.items()}
```

```python
# using zip 
days= ['sunday','monday','tuesday','wednesday','thursday','friday','saturday']
temp_C=[30.5,32.6,31.8,33.4,29.8,30.2,29.9]

{i:j for(i,j) in zip(days,temp_C)}
```

```python
# using if condition in zip
products ={'phone':10,'laptops':0,'charger':32,'tablet':0}
{key:value for (key,value) in products.items() if value>0}
```

```python
# Nested Comprehension
#print tables from numbers form 2 t0 4
{i:{j:i*j for j in range (1,11)} for i in range (2,5)}
```
