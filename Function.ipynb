```python
def is_even(num):
    """
    This function returns if  a given number is odd or even
    input - any valid integer
    output - even
    """
    if num%2==0:
        return 'even'
    else:
        return 'odd'
for i in range(1,11):
    x=is_even(i)
    print(x)
```

## Types of Arguments
* Default Argument
* Positional Argument
* Keyword Argument

```python
def power(a=1,b=1): # Default Argument 
    return a**b
power()
```

```python
# Positional Argument --> default behaior of function is called positional Argument 
                        # first argument goes to first parameter and second argument goes to second parameter
power(2,3) 
```

```python
# key Argument
power(b=3,a=2)
```

## *args and **kwargs
####  * args and  **kwargs are special keywords that are used to pass the variable length of arguments to a function.

```python
# *args 
# allows us to pass a variable number og non keyword arguments to a function.
def multiply1(a,b,c):
    return a*b*c
multiply1(2,3,4) 
```

```python
# in *arg methods
def multiply2(*args):
    product=1
    for i in args:
        product=product*i
    return product
multiply2(1,2,3,4,5,6,7)
```

```python
# **kwargs
# **Kwargs allow us to pass any number of keyword arguments.
# Keyword arguments mean that they contain a key-value pair, like a dictionary
def display(**kwargs):
    for(key,value) in kwargs.items():
        print(key ,'->',value)
display(Pakistan='Islamabad',Iran='Tehran',Saudi_Arabia = 'Raiz',India='delhi')
```
