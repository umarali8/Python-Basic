### Exceptions

If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happened.

- Exceptions are raised by python runtime
- You have to takle is on the fly

#### **Examples**

- Memory overflow
- Divide by 0 -> logical error
- Database error

```python
# catching specific exception
try:
  m=5
  f = open('sample.txt','r')
  print(f.read())
  print(m)
  print(5/0)
except FileNotFoundError:
  print('file not found')
except NameError:
  print('variable not defined')
except ZeroDivisionError:
  print("can't divide by 0")
except Exception as e:
  print(e)
```

```python
# else 
try: 
    f=open('sample.txt','r')
except FileNotFoundError:
    print('File Not Found')
except Exception:
    print('Is any problem')
else: # if try block does not have any problem then else block automatically run 
      # otherwise the except will execute
    print(f.read())
    
```

```python
# finally
try: 
    f=open('sample.txt','r')
except FileNotFoundError:
    print('File Not Found')
except Exception:
    print('Is any problem')
else: # if try block does not have any problem then else block automatically run 
      # otherwise the except will execute
    print(f.read())
finally:
    print('ye to print hoga hi')
```

```python
class Bank:
  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise Exception('amount cannot be -ve')
    if self.balance < amount:
      raise Exception('Insufficient amount')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(15000)
except Exception as e:
  print(e)
else:
  print(obj.balance)
```

```python
# creating custom exceptions
# exception hierarchy in python
class MyException(Exception):
  def __init__(self,message):
    print(message)

class Bank:

  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise MyException('amount cannot be -ve')
    if self.balance < amount:
      raise MyException('Insufficient amount')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(15000)
except MyException as e:
  pass
else:
  print(obj.balance)
    
```

```python
class SecurityError(Exception):

  def __init__(self,message):
    print(message)

  def logout(self):
    print('logout')

class Google:

  def __init__(self,name,email,password,device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self,email,password,device):
    if device != self.device:
      raise SecurityError('bhai teri to lag gayi')
    if email == self.email and password == self.password:
      print('welcome')
    else:
      print('login error')



obj = Google('umar','umar@gmail.com','1234','android')

try:
  obj.login('umar@gmail.com','1234','android')
except SecurityError as e:
  e.logout()
else:
  print(obj.name)
finally:
  print('database connection closed')
```
